````python
from mlx_lm import load, stream_generate
from mlx_lm.sample_utils import make_sampler

user_supermarket_query = "What is the cheaper in tesco, ground cumin or whole cumin"
system_prompt = "You summarise Irish supermarket product query results in a concise, helpful way." \
    "When comparing product prices pay particular attention to unit prices vs actual price and explore both options if not specified by the user, all prices are in euros. Base your answer on the actual data, not assumptions. " \
    "When comparing value always use the unit price. If the title of the product is unclear refer to the description to resolve any ambiguity"
prompt = f"User asked: {user_supermarket_query}? \n\nQuery results:\n{results_markdown_kv}"

# Apply the chat template for Qwen
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": prompt }]
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True,enable_thinking=False)

sampler = make_sampler(temp=0.7, top_p=0.9, top_k = 20)

# 5. Run the stream
for response in stream_generate(
    model,
    tokenizer,
    prompt,
    max_tokens=4096,
    sampler=sampler  # Pass the sampler here
):
    print(response.text, end="", flush=True)
````
